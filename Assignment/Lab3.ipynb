```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105],
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Edward"],
    "Department": ["HR", "IT", "IT", "Marketing", "Sales"],
    "Age": [29, 34, 41, 28, 38],
    "Salary": [50000, 70000, 65000, 55000, 60000],
    "Years_of_Experience": [4, 8, 10, 3, 12],
    "Joining_Date": ["2020-03-15", "2017-07-19", "2013-06-01", "2021-02-10", "2010-11-25"],
    "Gender": ["Female", "Male", "Male", "Female", "Male"],
    "Bonus": [5000, 7000, 6000, 4500, 5000],
    "Rating": [4.5, 4.0, 3.8, 4.7, 3.5]
}

df_initial = pd.DataFrame(data)
df_initial.to_csv("employees.csv", index=False)
df = pd.read_csv("employees.csv")


print("a)Shape:-")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}\n")

print("b) DataFrame Summary ")
df.info()

print("c) Descriptive Statistics")
print(df.describe(), "\n")
print("d) i First 5 Rows ")
print(df.head(5))

print("d) iiLast 3 Rows ")
print(df.tail(3), "\n")


print("e) Statistics")
print("i.   Average Salary:", df["Salary"].mean())
print("ii.  Total Bonus Paid:", df["Bonus"].sum())
print("iii. Youngest Employee Age:", df["Age"].min())
print("iv.  Highest Performance Rating:", df["Rating"].max(), "\n")

print(" f) Sorted by Salary (Descending) ")
df_sorted = df.sort_values(by="Salary", ascending=False)
print(df_sorted, "\n")


def categorize_rating(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    else:
        return "Average"

df["Performance_Category"] = df["Rating"].apply(categorize_rating)
print("g) Added Performance Category Column ")
print(df[["Name", "Rating", "Performance_Category"]], "\n")

print("h) Missing Values Count per Column ")
print(df.isnull().sum(), "\n")

df.rename(columns={"Employee_ID": "ID"}, inplace=True)
print("i) Renamed 'Employee_ID' to 'ID' ")
print(df.columns.tolist(), "\n")


print("j.i) Employees with > 5 Years Experience ")
print(df[df["Years_of_Experience"] > 5][["ID", "Name", "Years_of_Experience"]])

print("\nj.ii) Employees in IT Department ---")
print(df[df["Department"] == "IT"][["ID", "Name", "Department"]], "\n")

df["Tax"] = df["Salary"] * 0.10
print(" k) Added Tax Column (10% of Salary) ")
print(df[["ID", "Name", "Salary", "Tax"]], "\n")

df.to_csv("modified_employees.csv", index=False)
print(" l) Saved modified DataFrame to 'modified_employees.csv' successfully! ")
```
