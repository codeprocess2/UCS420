```python
import numpy as np
```

```python

arr=np.array([10,20,30,40,50])
print(arr+2)
print(arr*3)
print(arr/2)
```

```python
arr=np.array([1, 2, 3, 6, 4, 5])
print(arr[::-1])
```

```python
x=np.array([1,2,3,4,5,1,2,1,1,1])
values, counts=np.unique(x, return_counts=True)
freq=values[np.argmax(counts)]
print("Most frequent value ", freq)
print("Indices ",np.where(x == freq)[0])
```

```python
y=np.array([1,1,1,2,3,4,2,4,3,3])
values, counts=np.unique(y, return_counts=True)
freq=values[np.argmax(counts)]
print("Most frequent value:", freq)
print("Indices:", np.where(y == freq)[0])
```

```python
arr=np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])

print(arr[0,1])
print(arr[2,0])
```

```python
Vansh=np.linspace(10, 100, 25)
print(Vansh)
print("Dimensions:", Vansh.ndim)
print("Shape:", Vansh.shape)
print("Total elements:", Vansh.size)
print("Data type:", Vansh.dtype)
print("Total bytes:", Vansh.nbytes)
transpose = Vansh.reshape(25,1)
print(transpose)
print(Vansh.T)
print("For a 1-D array, T does not change its shape.")
```

```python
ucs420_vansh = np.array([[10,20,30,40],
                         [50,60,70,80],
                         [90,15,20,35]])

print("Mean:", np.mean(ucs420_vansh))
print("Median:", np.median(ucs420_vansh))
print("Maximum:", np.max(ucs420_vansh))
print("Minimum:", np.min(ucs420_vansh))
print("Unique elements:", np.unique(ucs420_vansh))
reshaped_ucs420_vansh = ucs420_vansh.reshape(4,3)
print(reshaped_ucs420_vansh)
resized_ucs420_vansh = np.resize(ucs420_vansh, (2,3))
print(resized_ucs420_vansh)
```
