```python
import numpy as np

temperature=np.array([25, 28, 31, 35, 38, 27, 33, 40])
temperature=temperature + 2
fahrenheit=(9/5) * temperature + 32
greater=temperature[temperature > 32]
count=np.sum(temperature > 32)
print("Updated temperatures:", temperature)
print("Fahrenheit:", fahrenheit)
print("Readings greater than 32:", greater)
print("Count:", count)
```

```python
steps=np.array([[5000, 6200, 7100],
                  [8000, 7500, 9000],
                  [4500, 5100, 4800],
                  [9000, 8500, 9500]])
print("Total steps:", np.sum(steps))
print("Mean steps:", np.mean(steps))
print("Maximum:", np.max(steps))
print("Minimum:", np.min(steps))
print("Total for each day:", np.sum(steps, axis=0))
print("Total for each user:", np.sum(steps, axis=1))
pos=np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position:", pos)
```

```python
matrix = np.arange(1,13).reshape(3,4)
print(matrix)
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:,1])
print("Rows 1-2 and columns 2-3:", matrix[1:3,2:4])
```

```python
matrix=np.arange(1,13).reshape(3,4)
flat_array=matrix.flatten()
ravel_array=matrix.ravel()
print("Flatten:",flat_array)
print("Ravel:",ravel_array)
ravel_array[0]=100
print("After ravel change:")
print(matrix)
print(ravel_array)
flat_array[1]=200
print("After flatten change:")
print(matrix)
print(flat_array)
```

```python

X=np.array([[6,70,3],
              [5,50,6],
              [8,80,2],
              [4,30,8]])

y=np.array([40,65,30,85])

beta=np.linalg.inv(X.T @ X) @ X.T @ y

print("Coefficients:", beta)

new_user=np.array([5,40,7])

prediction=new_user@beta
print("Predicted assistance score:", prediction)
```
