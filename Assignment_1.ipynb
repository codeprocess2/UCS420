```python


n = int(input("Enter value of n: "))

sum = 0

for num in range(2, n + 1):
    prime=True
    
    for i in range(2, int(num ** 0.5) + 1):
        if num%i==0:
            prime=False
            break
    
    if prime:
        sum += num

print("Sum of prime numbers:", sum)
```

```python
def IsPrime(n):
	for i in range(2, n//2 + 1):
		if n%i==0:
			return 0
	return 1

print ("IsPrime 20  --> ", IsPrime(20))
print ("IsPrime 23  --> ", IsPrime(23))
print ("IsPrime 200 --> ", IsPrime(200))
print ("IsPrime(37)  --> ", IsPrime(37))
     
```

```python
def AddN(n):
	s= sum(range(n+1))
	return s

print ("AddN(10)  --> ", AddN(10))
print ("AddN(20)  --> ", AddN(20))
print ("AddN(50)  --> ", AddN(50))
print ("AddN(200) --> ", AddN(200))
```

```python
text = "Indian Army"

print("Original--> ", text)
print("Step +1 --> ", text[::1])
print("Step +2 --> ", text[::2])
print("Reversed --> ", text[::-1])
print("Reverse step 2 --> ", text[::-2])

text = text[::-1]
print("After reverse  --> ", text)
```
